```python
import numpy as np 
import pandas as pd 

print(f"Pandas version: {pd.__version__}")

# Creating Series 
cities = pd.Series(['San Francisco', 'San Jose', 'Sacramento'])
population = pd.Series([852469, 1015785, 485199]) 
# Combining into a DataFrame 
city_df = pd.DataFrame({'City Name': cities, 'Population': population}) 
city_df
```

```python
# Load dataset from url 
url = "https://download.mlcc.google.com/mledu-datasets/california_housing_train.csv"
housing_df = pd.read_csv(url)
# Inspect statistics
print("First 5 rows:") 
display(housing_df.head()) 
print("\\nSummary Statistics:") 
display(housing_df.describe())
```

```python
#Add derived feature: Rooms per house 
housing_df['rooms_per_house'] = (
    housing_df['total_rooms'] / housing_df['households']
)
#Simple filtering: Median house value &gt; 200,000 
high_value_df = housing_df[housing_df['median_house_value'] > 200000] 
high_value_df.head()
```
